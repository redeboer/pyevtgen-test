`phase_space()` generates uniform phase space for arbitrary masses. It needs no `Generator`, so it starts instantly and can be called with different masses as often as you like. Change the masses below and rerun the cell:

```python
import matplotlib.pyplot as plt
import numpy as np

from pyevtgen import phase_space

parent_mass = 3.0969
masses = [0.1350, 0.1396, 0.4937]
momenta = phase_space(parent_mass, masses, n=100_000, seed=0)


def mass_squared(p: np.ndarray) -> np.ndarray:
    return p[..., 0] ** 2 - np.sum(p[..., 1:] ** 2, axis=-1)


fig, ax = plt.subplots(figsize=(4.5, 4))
ax.hist2d(
    mass_squared(momenta[:, 0] + momenta[:, 1]),
    mass_squared(momenta[:, 1] + momenta[:, 2]),
    bins=60,
    cmin=1,
)
ax.set_xlabel("$m_{01}^2$ [GeV$^2$]")
ax.set_ylabel("$m_{12}^2$ [GeV$^2$]")
plt.show()
```

With more daughters, a single Dalitz plot no longer shows all degrees of freedom. The invariant masses of all pairs give an overview, for instance for five pions:

```python
from itertools import combinations

momenta = phase_space(3.0969, [0.1396] * 5, n=50_000, seed=0)
fig, ax = plt.subplots(figsize=(6, 3.5))
for i, j in combinations(range(5), 2):
    pair = momenta[:, i] + momenta[:, j]
    ax.hist(np.sqrt(mass_squared(pair)), bins=60, histtype="step")
ax.set_xlabel(r"$m(\pi\pi)$ [GeV]")
plt.show()
```
