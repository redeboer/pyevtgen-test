# Particle properties

The particle table defines the names, PDG IDs, masses, widths and lifetimes that EvtGen uses. A {py:class}`.Generator` gives access to it.

```python
import pandas as pd
```

Creating a {py:class}`.Generator` reads the decay and particle tables and initializes Pythia8, which prints a lot of log output. That output is hidden on this page.

```python
from pyevtgen import Generator

generator = Generator()
```

{py:meth}`.Generator.properties` looks up a particle by EvtGen name or PDG ID and returns its {py:class}`.ParticleProperties`:

```python
generator.properties("J/psi")
```

```python
properties = generator.properties(443)
properties.mass, properties.width, properties.spin2
```

Spins are stored as twice the spin, {py:attr}`~.ParticleProperties.spin2`, so that they are integers. Masses and widths are in GeV and {py:attr}`~.ParticleProperties.ctau` in mm. The mass range from {py:attr}`~.ParticleProperties.min_mass` to {py:attr}`~.ParticleProperties.max_mass` limits the mass that EvtGen samples for a resonance.

## The full table

{py:meth}`.Generator.particles` returns all entries, which is convenient to search the table:

```python
table = pd.DataFrame([
    {
        "name": p.name,
        "PDG ID": p.pdg_id,
        "mass": p.mass,
        "width": p.width,
        "ctau": p.ctau,
        "charge": p.charge,
        "spin2": p.spin2,
    }
    for p in generator.particles()
])
len(table)
```

```python
table[table["name"].str.startswith("D_s")]
```

Unknown particles raise a {py:class}`KeyError`:

```python
try:
    generator.properties("D_x+")
except KeyError as error:
    print(error)
```
